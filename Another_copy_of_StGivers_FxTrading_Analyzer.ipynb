```python

# ===== GIVERS AI: DATA PREPARATION =====

def prepare_data(df):
    df = df.copy()

    if isinstance(df.columns, pd.MultiIndex):
        df.columns = df.columns.get_level_values(0)

    df = df.loc[:, ~df.columns.duplicated()]

    required = ["Open", "High", "Low", "Close"]
    df = df.dropna(subset=required)

    for col in required:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    df = df.dropna(subset=required)

    return df

m15 = prepare_data(frames["M15"])
h1 = prepare_data(frames["H1"])

print("===== GIVERS AI PREPARATION =====")
print("M15 ready:", len(m15))
print("H1 ready:", len(h1))
print("M15 columns:", m15.columns.tolist())
print("H1 columns:", h1.columns.tolist())

print("\nDATA PREPARATION COMPLETE")
```

```python

# ===== GIVERS AI: H1 TREND ENGINE =====

h1["EMA_20"] = h1["Close"].ewm(span=20, adjust=False).mean()
h1["EMA_50"] = h1["Close"].ewm(span=50, adjust=False).mean()

def detect_trend(row):
    if row["EMA_20"] > row["EMA_50"]:
        return "BULLISH"
    elif row["EMA_20"] < row["EMA_50"]:
        return "BEARISH"
    return "SIDEWAYS"

h1["H1_Trend"] = h1.apply(detect_trend, axis=1)

print("===== GIVERS AI H1 TREND REPORT =====")
print(h1[["Close", "EMA_20", "EMA_50", "H1_Trend"]].tail(10))

print("\nLATEST H1 TREND:", h1["H1_Trend"].iloc[-1])
print("\nH1 TREND ENGINE COMPLETE")
```

```python

# ===== GIVERS AI: M15 TREND ENGINE =====

m15["EMA_20"] = m15["Close"].ewm(
    span=20, adjust=False
).mean()

m15["EMA_50"] = m15["Close"].ewm(
    span=50, adjust=False
).mean()

m15["M15_Trend"] = np.where(
    m15["EMA_20"] > m15["EMA_50"], "BULLISH",
    np.where(
        m15["EMA_20"] < m15["EMA_50"],
        "BEARISH", "SIDEWAYS"
    )
)

print("===== GIVERS AI M15 TREND REPORT =====")
print(m15[[
    "Close", "EMA_20", "EMA_50", "M15_Trend"
]].tail(10))

print("\nLATEST M15 TREND:", m15["M15_Trend"].iloc[-1])
print("\nM15 TREND ENGINE COMPLETE")
```

```python

# ===== GIVERS AI: MULTI-TIMEFRAME ENGINE =====

h1_trend = h1["H1_Trend"].iloc[-1]
m15_trend = m15["M15_Trend"].iloc[-1]

if h1_trend == "BULLISH" and m15_trend == "BULLISH":
    decision = "BUY BIAS"

elif h1_trend == "BEARISH" and m15_trend == "BEARISH":
    decision = "SELL BIAS"

else:
    decision = "NO TRADE"

print("===== GIVERS AI DIRECTIONAL REPORT =====")
print("H1 TREND:", h1_trend)
print("M15 TREND:", m15_trend)
print("----------------------------------------")
print("DIRECTION:", decision)
print("----------------------------------------")
print("MULTI-TIMEFRAME ENGINE COMPLETE")
```

```python

# ===== GIVERS AI: LIQUIDITY SWEEP ENGINE =====

m15["Previous_High"] = m15["High"].shift(1).rolling(10).max()
m15["Previous_Low"] = m15["Low"].shift(1).rolling(10).min()

m15["Liquidity"] = "NONE"

buy_sweep = (
    (m15["Low"] < m15["Previous_Low"]) &
    (m15["Close"] > m15["Previous_Low"])
)

sell_sweep = (
    (m15["High"] > m15["Previous_High"]) &
    (m15["Close"] < m15["Previous_High"])
)

m15.loc[buy_sweep, "Liquidity"] = "BULLISH SWEEP"
m15.loc[sell_sweep, "Liquidity"] = "BEARISH SWEEP"

print("===== GIVERS AI LIQUIDITY REPORT =====")
print(m15[[
    "Close", "Previous_High",
    "Previous_Low", "Liquidity"
]].tail(15))

print("\nLATEST LIQUIDITY:", m15["Liquidity"].iloc[-1])
print("\nLIQUIDITY ENGINE COMPLETE")
```

```python

# ===== GIVERS AI: DISPLACEMENT ENGINE =====

m15["Candle_Body"] = abs(
    m15["Close"] - m15["Open"]
)

m15["Average_Body"] = (
    m15["Candle_Body"]
    .rolling(20)
    .mean()
    .shift(1)
)

m15["Displacement"] = "NONE"

bullish_displacement = (
    (m15["Close"] > m15["Open"]) &
    (m15["Candle_Body"] >= 1.5 * m15["Average_Body"])
)

bearish_displacement = (
    (m15["Close"] < m15["Open"]) &
    (m15["Candle_Body"] >= 1.5 * m15["Average_Body"])
)

m15.loc[
    bullish_displacement,
    "Displacement"
] = "BULLISH DISPLACEMENT"

m15.loc[
    bearish_displacement,
    "Displacement"
] = "BEARISH DISPLACEMENT"

print("===== GIVERS AI DISPLACEMENT REPORT =====")

print(m15[[
    "Close",
    "Candle_Body",
    "Average_Body",
    "Displacement"
]].tail(15))

print("\nLATEST DISPLACEMENT:",
      m15["Displacement"].iloc[-1])

print("\nDISPLACEMENT ENGINE COMPLETE")
```

```python

# ===== GIVERS AI: BOS ENGINE =====

m15["Structure_High"] = (
    m15["High"].shift(1).rolling(10).max()
)

m15["Structure_Low"] = (
    m15["Low"].shift(1).rolling(10).min()
)

m15["BOS"] = "NONE"

bullish_bos = (
    m15["Close"] > m15["Structure_High"]
)

bearish_bos = (
    m15["Close"] < m15["Structure_Low"]
)

m15.loc[bullish_bos, "BOS"] = "BULLISH BOS"
m15.loc[bearish_bos, "BOS"] = "BEARISH BOS"

print("===== GIVERS AI BOS REPORT =====")

print(m15[[
    "Close",
    "Structure_High",
    "Structure_Low",
    "BOS"
]].tail(15))

print("\nLATEST BOS:", m15["BOS"].iloc[-1])
print("\nBOS ENGINE COMPLETE")
```

```python

# ===== GIVERS AI: FVG ENGINE =====

m15["FVG"] = "NONE"

bullish_fvg = (
    m15["Low"] >
    m15["High"].shift(2)
)

bearish_fvg = (
    m15["High"] <
    m15["Low"].shift(2)
)

m15.loc[bullish_fvg, "FVG"] = "BULLISH FVG"
m15.loc[bearish_fvg, "FVG"] = "BEARISH FVG"

print("===== GIVERS AI FVG REPORT =====")

print(m15[[
    "Close",
    "High",
    "Low",
    "FVG"
]].tail(20))

print("\nLATEST FVG:", m15["FVG"].iloc[-1])

print("\nFVG ENGINE COMPLETE")
```

```python

# ===== GIVERS AI: HYBRID CONFLUENCE SCANNER =====

import pandas as pd
import numpy as np

now = pd.Timestamp.now(tz="UTC")

# Keep only fully closed candles
m15_closed = m15.loc[
    m15.index + pd.Timedelta(minutes=15) <= now
].copy()

h1_closed = h1.loc[
    h1.index + pd.Timedelta(hours=1) <= now
].copy()

if len(m15_closed) < 60 or len(h1_closed) < 50:
    raise ValueError("Not enough closed candles.")

# Recalculate timeframe trends
h1_closed["EMA_20"] = h1_closed["Close"].ewm(
    span=20, adjust=False
).mean()

h1_closed["EMA_50"] = h1_closed["Close"].ewm(
    span=50, adjust=False
).mean()

m15_closed["EMA_20"] = m15_closed["Close"].ewm(
    span=20, adjust=False
).mean()

m15_closed["EMA_50"] = m15_closed["Close"].ewm(
    span=50, adjust=False
).mean()

h1_trend = (
    "BULLISH" if h1_closed["EMA_20"].iloc[-1] >
    h1_closed["EMA_50"].iloc[-1] else
    "BEARISH" if h1_closed["EMA_20"].iloc[-1] <
    h1_closed["EMA_50"].iloc[-1] else "SIDEWAYS"
)

m15_trend = (
    "BULLISH" if m15_closed["EMA_20"].iloc[-1] >
    m15_closed["EMA_50"].iloc[-1] else
    "BEARISH" if m15_closed["EMA_20"].iloc[-1] <
    m15_closed["EMA_50"].iloc[-1] else "SIDEWAYS"
)

# Liquidity sweeps
m15_closed["Prev_High"] = (
    m15_closed["High"].shift(1).rolling(10).max()
)

m15_closed["Prev_Low"] = (
    m15_closed["Low"].shift(1).rolling(10).min()
)

m15_closed["Liquidity"] = "NONE"

m15_closed.loc[
    (m15_closed["Low"] < m15_closed["Prev_Low"]) &
    (m15_closed["Close"] > m15_closed["Prev_Low"]),
    "Liquidity"
] = "BULLISH SWEEP"

m15_closed.loc[
    (m15_closed["High"] > m15_closed["Prev_High"]) &
    (m15_closed["Close"] < m15_closed["Prev_High"]),
    "Liquidity"
] = "BEARISH SWEEP"

# Displacement
body = (m15_closed["Close"] - m15_closed["Open"]).abs()
avg_body = body.rolling(20).mean().shift(1)

m15_closed["Displacement"] = "NONE"

m15_closed.loc[
    (m15_closed["Close"] > m15_closed["Open"]) &
    (body >= 1.5 * avg_body),
    "Displacement"
] = "BULLISH DISPLACEMENT"

m15_closed.loc[
    (m15_closed["Close"] < m15_closed["Open"]) &
    (body >= 1.5 * avg_body),
    "Displacement"
] = "BEARISH DISPLACEMENT"

# BOS
m15_closed["Structure_High"] = (
    m15_closed["High"].shift(1).rolling(10).max()
)

m15_closed["Structure_Low"] = (
    m15_closed["Low"].shift(1).rolling(10).min()
)

m15_closed["BOS"] = "NONE"

m15_closed.loc[
    m15_closed["Close"] > m15_closed["Structure_High"],
    "BOS"
] = "BULLISH BOS"

m15_closed.loc[
    m15_closed["Close"] < m15_closed["Structure_Low"],
    "BOS"
] = "BEARISH BOS"

# FVG
m15_closed["FVG"] = "NONE"

m15_closed.loc[
    m15_closed["Low"] > m15_closed["High"].shift(2),
    "FVG"
] = "BULLISH FVG"

m15_closed.loc[
    m15_closed["High"] < m15_closed["Low"].shift(2),
    "FVG"
] = "BEARISH FVG"

# Check recent 3 closed candles for confirmation
recent = m15_closed.tail(3)

buy_confirmations = [
    recent["Liquidity"].eq("BULLISH SWEEP").any(),
    recent["Displacement"].eq("BULLISH DISPLACEMENT").any(),
    recent["BOS"].eq("BULLISH BOS").any(),
    recent["FVG"].eq("BULLISH FVG").any()
]

sell_confirmations = [
    recent["Liquidity"].eq("BEARISH SWEEP").any(),
    recent["Displacement"].eq("BEARISH DISPLACEMENT").any(),
    recent["BOS"].eq("BEARISH BOS").any(),
    recent["FVG"].eq("BEARISH FVG").any()
]

if h1_trend == "BULLISH" and m15_trend == "BULLISH" and all(buy_confirmations):
    decision = "BUY SETUP CANDIDATE"

elif h1_trend == "BEARISH" and m15_trend == "BEARISH" and all(sell_confirmations):
    decision = "SELL SETUP CANDIDATE"

else:
    decision = "NO TRADE"

print("===== GIVERS AI HYBRID REPORT =====")
print("H1 TREND:", h1_trend)
print("M15 TREND:", m15_trend)
print("CLOSED M15 CANDLES:", len(m15_closed))
print("LAST CLOSED CANDLE:", m15_closed.index[-1])

print("\nBUY CONFIRMATIONS:", sum(buy_confirmations), "/ 4")
print("SELL CONFIRMATIONS:", sum(sell_confirmations), "/ 4")

print("\nLATEST LIQUIDITY:", m15_closed["Liquidity"].iloc[-1])
print("LATEST DISPLACEMENT:", m15_closed["Displacement"].iloc[-1])
print("LATEST BOS:", m15_closed["BOS"].iloc[-1])
print("LATEST FVG:", m15_closed["FVG"].iloc[-1])

print("\n==================================")
print("GIVERS AI DECISION:", decision)
print("==================================")
```

```python

# GIVERS AI — SAFE INDICATOR ENGINE

def calculate_safe_indicators(df):
    df = df.copy()

    df["EMA_20"] = df["Close"].ewm(
        span=20, adjust=False
    ).mean()

    df["EMA_50"] = df["Close"].ewm(
        span=50, adjust=False
    ).mean()

    df["Previous_High"] = (
        df["High"].shift(1).rolling(10).max()
    )

    df["Previous_Low"] = (
        df["Low"].shift(1).rolling(10).min()
    )

    df["Liquidity"] = "NONE"

    df.loc[
        (df["Low"] < df["Previous_Low"]) &
        (df["Close"] > df["Previous_Low"]),
        "Liquidity"
    ] = "BULLISH SWEEP"

    df.loc[
        (df["High"] > df["Previous_High"]) &
        (df["Close"] < df["Previous_High"]),
        "Liquidity"
    ] = "BEARISH SWEEP"

    df["Candle_Body"] = abs(
        df["Close"] - df["Open"]
    )

    df["Average_Body"] = (
        df["Candle_Body"].rolling(20).mean().shift(1)
    )

    df["Displacement"] = "NONE"

    strong = (
        df["Candle_Body"] >= df["Average_Body"] * 1.5
    )

    df.loc[
        strong & (df["Close"] > df["Open"]),
        "Displacement"
    ] = "BULLISH DISPLACEMENT"

    df.loc[
        strong & (df["Close"] < df["Open"]),
        "Displacement"
    ] = "BEARISH DISPLACEMENT"

    df["Structure_High"] = (
        df["High"].shift(1).rolling(10).max()
    )

    df["Structure_Low"] = (
        df["Low"].shift(1).rolling(10).min()
    )

    df["BOS"] = "NONE"

    df.loc[
        df["Close"] > df["Structure_High"],
        "BOS"
    ] = "BULLISH BOS"

    df.loc[
        df["Close"] < df["Structure_Low"],
        "BOS"
    ] = "BEARISH BOS"

    df["FVG"] = "NONE"

    df.loc[
        df["Low"] > df["High"].shift(2),
        "FVG"
    ] = "BULLISH FVG"

    df.loc[
        df["High"] < df["Low"].shift(2),
        "FVG"
    ] = "BEARISH FVG"

    return df


# Calculate indicators independently in each segment
indicator_results = []

for sid, group in segments_new.items():
    calculated = calculate_safe_indicators(group)
    calculated["Segment"] = sid
    indicator_results.append(calculated)

m15_engine_new = pd.concat(
    indicator_results
).sort_index()

required = [
    "EMA_20",
    "EMA_50",
    "Previous_High",
    "Previous_Low",
    "Average_Body",
    "Structure_High",
    "Structure_Low"
]

m15_ready_new = m15_engine_new.dropna(
    subset=required
).copy()

print("===== INDICATOR ENGINE REPORT =====")
print("Safe candles:", len(m15_engine_new))
print("Ready candles:", len(m15_ready_new))
print("Segments:", m15_ready_new["Segment"].nunique())
print("Indicator calculation: COMPLETE")
```

```python

# GIVERS AI — CONFIRMED STRUCTURE ENGINE

def add_confirmed_structure(df, lookback=3):
    df = df.copy()
    results = []

    for segment_id, group in df.groupby("Segment", sort=False):
        group = group.sort_index().copy()

        window = (lookback * 2) + 1

        rolling_high = group["High"].rolling(
            window=window,
            center=True,
            min_periods=window
        ).max()

        rolling_low = group["Low"].rolling(
            window=window,
            center=True,
            min_periods=window
        ).min()

        swing_high = group["High"].where(
            group["High"] == rolling_high
        ).shift(lookback)

        swing_low = group["Low"].where(
            group["Low"] == rolling_low
        ).shift(lookback)

        group["Confirmed_Swing_High"] = swing_high
        group["Confirmed_Swing_Low"] = swing_low

        group["Last_Swing_High"] = swing_high.ffill()
        group["Last_Swing_Low"] = swing_low.ffill()

        results.append(group)

    return pd.concat(results).sort_index()


m15_structure_new = add_confirmed_structure(
    m15_ready_new,
    lookback=3
)

print("===== CONFIRMED STRUCTURE REPORT =====")
print("Candles:", len(m15_structure_new))
print(
    "Confirmed swing highs:",
    m15_structure_new["Confirmed_Swing_High"].notna().sum()
)
print(
    "Confirmed swing lows:",
    m15_structure_new["Confirmed_Swing_Low"].notna().sum()
)
print("Structure engine: RESTORED")
```

```python

# GIVERS AI — BOS V2 ENGINE

def build_bos_v2(df):
    output = df.copy()
    output["BOS_V2"] = "NONE"
    output["BOS_V2_Level"] = np.nan

    for segment_id, group in output.groupby("Segment", sort=False):
        group = group.sort_index()

        active_high = None
        active_low = None

        for position in range(len(group)):
            timestamp = group.index[position]
            row = group.iloc[position]

            if pd.notna(row["Confirmed_Swing_High"]):
                active_high = {
                    "level": float(row["Confirmed_Swing_High"]),
                    "broken": False
                }

            if pd.notna(row["Confirmed_Swing_Low"]):
                active_low = {
                    "level": float(row["Confirmed_Swing_Low"]),
                    "broken": False
                }

            if position == 0:
                continue

            previous_close = float(
                group.iloc[position - 1]["Close"]
            )
            close = float(row["Close"])

            if active_high is not None and not active_high["broken"]:
                level = active_high["level"]

                if previous_close <= level and close > level:
                    output.loc[timestamp, "BOS_V2"] = "BULLISH BOS"
                    output.loc[timestamp, "BOS_V2_Level"] = level
                    active_high["broken"] = True

            if active_low is not None and not active_low["broken"]:
                level = active_low["level"]

                if previous_close >= level and close < level:
                    output.loc[timestamp, "BOS_V2"] = "BEARISH BOS"
                    output.loc[timestamp, "BOS_V2_Level"] = level
                    active_low["broken"] = True

    return output


m15_structure_v2 = build_bos_v2(m15_structure_new)

print("===== BOS V2 REPORT =====")
print(
    "Total BOS:",
    (m15_structure_v2["BOS_V2"] != "NONE").sum()
)
print(
    "Bullish BOS:",
    (m15_structure_v2["BOS_V2"] == "BULLISH BOS").sum()
)
print(
    "Bearish BOS:",
    (m15_structure_v2["BOS_V2"] == "BEARISH BOS").sum()
)
```

```python

# GIVERS AI — SAFE SEQUENCE DETECTOR

def detect_safe_sequences(df):
    sequences = []

    for segment_id, group in df.groupby("Segment"):
        group = group.sort_index()

        for i in range(len(group)):
            row = group.iloc[i]
            direction = row["Trade_Direction"]

            if direction == "BUY":
                sweep_name = "BULLISH SWEEP"
                displacement_name = "BULLISH DISPLACEMENT"
                bos_name = "BULLISH BOS"
                fvg_name = "BULLISH FVG"

            elif direction == "SELL":
                sweep_name = "BEARISH SWEEP"
                displacement_name = "BEARISH DISPLACEMENT"
                bos_name = "BEARISH BOS"
                fvg_name = "BEARISH FVG"

            else:
                continue

            if row["Liquidity"] != sweep_name:
                continue

            displacement_i = None

            for j in range(i + 1, min(i + 4, len(group))):
                if group.iloc[j]["Displacement"] == displacement_name:
                    displacement_i = j
                    break

            if displacement_i is None:
                continue

            bos_i = None

            for j in range(
                displacement_i + 1,
                min(displacement_i + 4, len(group))
            ):
                if group.iloc[j]["BOS"] == bos_name:
                    bos_i = j
                    break

            if bos_i is None:
                continue

            fvg_i = None

            for j in range(
                bos_i + 1,
                min(bos_i + 4, len(group))
            ):
                if group.iloc[j]["FVG"] == fvg_name:
                    fvg_i = j
                    break

            if fvg_i is None:
                continue

            sequences.append({
                "Direction": direction,
                "Segment": segment_id,
                "Sweep_Time": group.index[i],
                "Displacement_Time": group.index[displacement_i],
                "BOS_Time": group.index[bos_i],
                "FVG_Time": group.index[fvg_i]
            })

    return sequences


print("Sequence detector restored successfully.")
```

```python

# GIVERS AI — FVG ENTRY VALIDATION V2

def find_safe_fvg_entry(sequence, df, max_wait=20):
    segment = df[df["Segment"] == sequence["Segment"]].sort_index()
    fvg_time = sequence["FVG_Time"]

    if fvg_time not in segment.index:
        return None

    i = segment.index.get_loc(fvg_time)

    if not isinstance(i, (int, np.integer)) or i < 2:
        return None

    candle = segment.iloc[i]
    previous = segment.iloc[i - 2]
    direction = sequence["Direction"]

    if direction == "SELL":
        zone_low = float(candle["High"])
        zone_high = float(previous["Low"])
    else:
        zone_low = float(previous["High"])
        zone_high = float(candle["Low"])

    if zone_low >= zone_high:
        return None

    future = segment.iloc[i + 1:i + 1 + max_wait]

    for entry_time, row in future.iterrows():
        touched = (
            float(row["Low"]) <= zone_high
            and float(row["High"]) >= zone_low
        )

        if direction == "SELL":
            confirmed = (
                row["Close"] < row["Open"]
                and row["Close"] < zone_low
            )
        else:
            confirmed = (
                row["Close"] > row["Open"]
                and row["Close"] > zone_high
            )

        if touched and confirmed:
            return {
                "Direction": direction,
                "FVG_Time": fvg_time,
                "Entry_Time": entry_time,
                "Zone_Low": zone_low,
                "Zone_High": zone_high,
                "Entry": float(row["Close"])
            }

    return None


validated_entries_v2 = []
rejected_entries_v2 = 0

for sequence in safe_sequences_v2:
    result = find_safe_fvg_entry(sequence, m15_ready_v2)

    if result is not None:
        validated_entries_v2.append(result)
    else:
        rejected_entries_v2 += 1

print("===== GIVERS AI ENTRY VALIDATION V2 =====")
print("Total sequences:", len(safe_sequences_v2))
print("Validated entries:", len(validated_entries_v2))
print("Rejected entries:", rejected_entries_v2)

for i, entry in enumerate(validated_entries_v2, start=1):
    print("\nENTRY:", i)
    print("Direction:", entry["Direction"])
    print("FVG time:", entry["FVG_Time"])
    print("Entry time:", entry["Entry_Time"])
    print("Entry price:", entry["Entry"])
    print("FVG zone:", entry["Zone_Low"], "to", entry["Zone_High"])
```

```python

# GIVERS AI — RISK CALCULATION V2

risk_reports_v2 = []

for entry in validated_entries_v2:
    sequence = next(
        s for s in safe_sequences_v2
        if s["FVG_Time"] == entry["FVG_Time"]
    )

    segment = m15_ready_v2[
        m15_ready_v2["Segment"] == sequence["Segment"]
    ].sort_index()

    risk_window = segment.loc[
        sequence["Sweep_Time"]:entry["Entry_Time"]
    ]

    entry_price = entry["Entry"]

    if entry["Direction"] == "SELL":
        stop_loss = float(risk_window["High"].max())
        risk = stop_loss - entry_price
        tp_1r = entry_price - risk
        tp_2r = entry_price - (2 * risk)
    else:
        stop_loss = float(risk_window["Low"].min())
        risk = entry_price - stop_loss
        tp_1r = entry_price + risk
        tp_2r = entry_price + (2 * risk)

    if risk <= 0:
        print("Invalid risk calculation:", entry["FVG_Time"])
        continue

    risk_reports_v2.append({
        "Direction": entry["Direction"],
        "Entry": entry_price,
        "Stop_Loss": stop_loss,
        "Risk": risk,
        "TP_1R": tp_1r,
        "TP_2R": tp_2r
    })

print("===== GIVERS AI RISK REPORT V2 =====")
print("Risk reports generated:", len(risk_reports_v2))

for i, report in enumerate(risk_reports_v2, start=1):
    print("\nTRADE:", i)
    for key, value in report.items():
        print(key + ":", value)
```

```python

# GIVERS AI — HISTORICAL BACKTEST V2

def test_trade(entry, risk_report, df, target_key):
    sequence = next(
        s for s in safe_sequences_v2
        if s["FVG_Time"] == entry["FVG_Time"]
    )

    segment = df[df["Segment"] == sequence["Segment"]].sort_index()

    future = segment.loc[
        segment.index > entry["Entry_Time"]
    ]

    stop = risk_report["Stop_Loss"]
    target = risk_report[target_key]

    for timestamp, candle in future.iterrows():
        hit_sl = candle["High"] >= stop
        hit_tp = candle["Low"] <= target

        if hit_sl and hit_tp:
            return "AMBIGUOUS", timestamp

        if hit_sl:
            return "LOSS", timestamp

        if hit_tp:
            return "WIN", timestamp

    return "NO HIT", None


backtest_results_v2 = []

for entry, risk_report in zip(
    validated_entries_v2, risk_reports_v2
):
    for target_key, rr_label in [
        ("TP_1R", "1:1"),
        ("TP_2R", "1:2")
    ]:
        outcome, exit_time = test_trade(
            entry,
            risk_report,
            m15_ready_v2,
            target_key
        )

        backtest_results_v2.append({
            "Direction": entry["Direction"],
            "Entry_Time": entry["Entry_Time"],
            "Risk_Reward": rr_label,
            "Outcome": outcome,
            "Exit_Time": exit_time
        })

print("===== GIVERS AI BACKTEST V2 =====")

for result in backtest_results_v2:
    print("\nEntry:", result["Entry_Time"])
    print("Risk/Reward:", result["Risk_Reward"])
    print("Outcome:", result["Outcome"])
    print("Exit time:", result["Exit_Time"])

print("\n===== BACKTEST SUMMARY =====")

for rr in ["1:1", "1:2"]:
    selected = [
        r for r in backtest_results_v2
        if r["Risk_Reward"] == rr
    ]

    wins = sum(r["Outcome"] == "WIN" for r in selected)
    losses = sum(r["Outcome"] == "LOSS" for r in selected)
    ambiguous = sum(r["Outcome"] == "AMBIGUOUS" for r in selected)
    no_hit = sum(r["Outcome"] == "NO HIT" for r in selected)

    print("\nRisk/Reward:", rr)
    print("Wins:", wins)
    print("Losses:", losses)
    print("Ambiguous:", ambiguous)
    print("No hit:", no_hit)
```

```python

# GIVERS AI — TRADING COST IMPACT TEST

PIP_SIZE = 0.0001  # EUR/USD
COST_SCENARIOS_PIPS = [0, 0.5, 1, 1.5, 2, 3]

print("===== GIVERS AI COST SENSITIVITY =====")

for rr in ["1:1", "1:2"]:
    print("\nRISK/REWARD:", rr)

    total_gross_r = 0
    total_risk_pips = 0
    trade_count = 0

    for result in backtest_results_v2:
        if result["Risk_Reward"] != rr:
            continue

        if result["Outcome"] not in ["WIN", "LOSS"]:
            continue

        entry = next(
            e for e in validated_entries_v2
            if e["Entry_Time"] == result["Entry_Time"]
        )

        risk_report = next(
            r for r in risk_reports_v2
            if r["Entry"] == entry["Entry"]
        )

        risk_pips = risk_report["Risk"] / PIP_SIZE

        if result["Outcome"] == "WIN":
            gross_r = 1 if rr == "1:1" else 2
        else:
            gross_r = -1

        total_gross_r += gross_r
        total_risk_pips += risk_pips
        trade_count += 1

    print("Trades tested:", trade_count)
    print("Gross result:", round(total_gross_r, 3), "R")

    for cost_pips in COST_SCENARIOS_PIPS:
        total_cost_r = sum(
            cost_pips / (
                r["Risk"] / PIP_SIZE
            )
            for r in risk_reports_v2
        )

        net_r = total_gross_r - total_cost_r

        print(
            "Round-trip cost:",
            cost_pips,
            "pips | Net result:",
            round(net_r, 3),
            "R"
        )

print("\nNOTE: Cost scenarios are illustrative,")
print("not confirmed Deriv trading costs.")
```

```python

# GIVERS AI — COST-AWARE PERFORMANCE V2

OBSERVED_SPREAD_PIPS = 1.0
SLIPPAGE_SCENARIOS = [0, 0.2, 0.5, 1.0]
PIP_SIZE = 0.0001

print("===== GIVERS AI COST-AWARE REPORT =====")
print("Observed spread:", OBSERVED_SPREAD_PIPS, "pips")
print("Commission: Not included")

for rr in ["1:1", "1:2"]:
    selected = [
        r for r in backtest_results_v2
        if r["Risk_Reward"] == rr
        and r["Outcome"] in ["WIN", "LOSS"]
    ]

    gross_r = 0
    total_cost_per_risk = 0

    for result in selected:
        entry = next(
            e for e in validated_entries_v2
            if e["Entry_Time"] == result["Entry_Time"]
        )

        risk_report = next(
            r for r in risk_reports_v2
            if r["Entry"] == entry["Entry"]
        )

        risk_pips = risk_report["Risk"] / PIP_SIZE

        trade_r = (
            1 if result["Outcome"] == "WIN" and rr == "1:1"
            else 2 if result["Outcome"] == "WIN"
            else -1
        )

        gross_r += trade_r
        total_cost_per_risk += OBSERVED_SPREAD_PIPS / risk_pips

    print("\nRISK/REWARD:", rr)
    print("Trades:", len(selected))
    print("Gross result:", round(gross_r, 3), "R")

    for slippage in SLIPPAGE_SCENARIOS:
        total_net_cost = sum(
            (OBSERVED_SPREAD_PIPS + slippage) /
            (risk_reports_v2[i]["Risk"] / PIP_SIZE)
            for i in range(len(risk_reports_v2))
        )

        net_r = gross_r - total_net_cost

        print(
            "Spread:", OBSERVED_SPREAD_PIPS,
            "| Slippage:", slippage,
            "| Estimated net:", round(net_r, 3), "R"
        )

print("\nNOTE: Illustrative cost adjustment only.")
print("Historical spread and execution are not yet modelled.")
```

```python

# GIVERS AI - STRATEGY ENGINE
# H1 = Main trend
# M15 = Entry timeframe

def givers_strategy(h1_trend, m15_trend, liquidity,
                    displacement, structure, ema_confirmation):

    score = 0
    reasons = []

    # H1 direction is compulsory
    if h1_trend not in ["BULLISH", "BEARISH"]:
        return "NO TRADE", "H1 trend is unclear"

    if h1_trend != m15_trend:
        return "NO TRADE", "H1 and M15 trends disagree"

    # Four additional checks
    checks = {
        "Liquidity sweep": liquidity,
        "Displacement": displacement,
        "Market structure": structure,
        "EMA confirmation": ema_confirmation
    }

    for name, result in checks.items():
        if result:
            score += 1
            reasons.append(name + ": CONFIRMED")

    if score >= 3:
        signal = "BUY" if h1_trend == "BULLISH" else "SELL"
    else:
        signal = "NO TRADE"

    return signal, f"Confirmations: {score}/4 | " + ", ".join(reasons)


print("GIVERS AI STRATEGY ENGINE READY!")
```

```python

def givers_get_trend(df):
    import pandas as pd

    if isinstance(df.columns, pd.MultiIndex):
        close_columns = [
            col for col in df.columns
            if "Close" in col
        ]
        if not close_columns:
            return "INSUFFICIENT DATA"
        close = df[close_columns[0]]
    else:
        close = df["Close"]

    if isinstance(close, pd.DataFrame):
        close = close.iloc[:, 0]

    close = pd.to_numeric(close, errors="coerce").dropna()

    if len(close) < 50:
        return "INSUFFICIENT DATA"

    ema20 = close.ewm(span=20, adjust=False).mean().iloc[-1]
    ema50 = close.ewm(span=50, adjust=False).mean().iloc[-1]

    if ema20 > ema50:
        return "BULLISH"
    elif ema20 < ema50:
        return "BEARISH"
    else:
        return "SIDEWAYS"


h1_trend = givers_get_trend(frames["H1"])
m15_trend = givers_get_trend(frames["M15"])

print("GIVERS AI - LIVE TREND ANALYSIS")
print("==============================")
print("H1 TREND:", h1_trend)
print("M15 TREND:", m15_trend)
```

```python

# GIVERS AI - LIQUIDITY SWEEP DETECTOR

def detect_liquidity_sweep(df):

    # Use the last CLOSED candle
    candles = df.iloc[:-1].copy()

    if len(candles) < 10:
        return "INSUFFICIENT DATA"

    previous = candles.iloc[-2]
    last = candles.iloc[-1]

    # Compare against the previous 5 candles
    previous_high = candles["High"].iloc[-7:-2].max()
    previous_low = candles["Low"].iloc[-7:-2].min()

    if last["High"] > previous_high and last["Close"] < previous_high:
        return "BEARISH LIQUIDITY SWEEP"

    elif last["Low"] < previous_low and last["Close"] > previous_low:
        return "BULLISH LIQUIDITY SWEEP"

    return "NO LIQUIDITY SWEEP"


liquidity_result = detect_liquidity_sweep(frames["M15"])

print("GIVERS AI LIQUIDITY ANALYSIS")
print("============================")
print(liquidity_result)
```

```python

# GIVERS AI - DISPLACEMENT DETECTOR

def detect_displacement(df):

    candles = df.iloc[:-1].copy()

    if len(candles) < 25:
        return "INSUFFICIENT DATA"

    # Last closed candle
    last = candles.iloc[-1]

    # Previous 20 candle bodies
    previous_bodies = (
        candles["Close"].iloc[-21:-1]
        - candles["Open"].iloc[-21:-1]
    ).abs()

    average_body = previous_bodies.mean()
    last_body = abs(last["Close"] - last["Open"])

    if average_body == 0:
        return "INSUFFICIENT DATA"

    if last_body >= average_body * 1.5:
        if last["Close"] > last["Open"]:
            return "BULLISH DISPLACEMENT"
        elif last["Close"] < last["Open"]:
            return "BEARISH DISPLACEMENT"

    return "NO STRONG DISPLACEMENT"


displacement_result = detect_displacement(frames["M15"])

print("GIVERS AI DISPLACEMENT ANALYSIS")
print("================================")
print(displacement_result)
```

```python

# GIVERS AI - MARKET STRUCTURE DETECTOR

def detect_structure(df):

    candles = df.iloc[:-1].copy()

    if len(candles) < 25:
        return "INSUFFICIENT DATA"

    last = candles.iloc[-1]

    previous_high = candles["High"].iloc[-21:-1].max()
    previous_low = candles["Low"].iloc[-21:-1].min()

    if last["Close"] > previous_high:
        return "BULLISH BOS"

    elif last["Close"] < previous_low:
        return "BEARISH BOS"

    return "NO STRUCTURE BREAK"


structure_result = detect_structure(frames["M15"])

print("GIVERS AI MARKET STRUCTURE")
print("==========================")
print(structure_result)
```

```python

# GIVERS AI - EMA CONFIRMATION

def detect_ema_confirmation(df):

    import pandas as pd

    candles = df.iloc[:-1].copy()
    close = candles["Close"]

    if isinstance(close, pd.DataFrame):
        close = close.iloc[:, 0]

    close = pd.to_numeric(close, errors="coerce").dropna()

    if len(close) < 50:
        return "INSUFFICIENT DATA"

    ema20 = close.ewm(span=20, adjust=False).mean().iloc[-1]
    ema50 = close.ewm(span=50, adjust=False).mean().iloc[-1]

    if ema20 < ema50:
        return "BEARISH EMA CONFIRMATION"

    elif ema20 > ema50:
        return "BULLISH EMA CONFIRMATION"

    return "NO EMA CONFIRMATION"


ema_result = detect_ema_confirmation(frames["M15"])

print("GIVERS AI EMA ANALYSIS")
print("======================")
print(ema_result)
```

```python

# GIVERS AI - ADVANCED MARKET STRUCTURE

def detect_advanced_structure(df):

    import pandas as pd

    candles = df.iloc[:-1].copy()

    if isinstance(candles.columns, pd.MultiIndex):
        candles.columns = candles.columns.get_level_values(0)

    required = ["High", "Low", "Close"]

    if not all(col in candles.columns for col in required):
        return "INSUFFICIENT DATA"

    candles = candles[required].dropna().tail(150)

    if len(candles) < 30:
        return "INSUFFICIENT DATA"

    # Confirmed swing points: 2 candles on each side
    swing_highs = []
    swing_lows = []

    for i in range(2, len(candles) - 2):
        high = candles["High"].iloc[i]
        low = candles["Low"].iloc[i]

        if high > candles["High"].iloc[i-2:i].max() and high > candles["High"].iloc[i+1:i+3].max():
            swing_highs.append((i, high))

        if low < candles["Low"].iloc[i-2:i].min() and low < candles["Low"].iloc[i+1:i+3].min():
            swing_lows.append((i, low))

    if len(swing_highs) < 2 or len(swing_lows) < 2:
        return "INSUFFICIENT SWING DATA"

    previous_high, latest_high = swing_highs[-2:]
    previous_low, latest_low = swing_lows[-2:]

    bullish_trend = (
        latest_high[1] > previous_high[1]
        and latest_low[1] > previous_low[1]
    )

    bearish_trend = (
        latest_high[1] < previous_high[1]
        and latest_low[1] < previous_low[1]
    )

    last_close = candles["Close"].iloc[-1]
    previous_close = candles["Close"].iloc[-2]

    if previous_close <= latest_high[1] and last_close > latest_high[1]:
        return "BEARISH CHoCH" if bearish_trend else "BULLISH BOS"

    if previous_close >= latest_low[1] and last_close < latest_low[1]:
        return "BULLISH CHoCH" if bullish_trend else "BEARISH BOS"

    return "NO STRUCTURE BREAK"


advanced_structure = detect_advanced_structure(frames["M15"])

print("GIVERS AI ADVANCED STRUCTURE")
print("============================")
print(advanced_structure)
```

```python

# GIVERS AI - FAIR VALUE GAP DETECTOR

def detect_fvg(df):

    candles = df.iloc[:-1].copy()

    if len(candles) < 3:
        return "INSUFFICIENT DATA"

    # Last three CLOSED candles
    first = candles.iloc[-3]
    third = candles.iloc[-1]

    # Bullish FVG: third candle's low is above first candle's high
    if third["Low"] > first["High"]:
        return {
            "type": "BULLISH FVG",
            "lower": float(first["High"]),
            "upper": float(third["Low"])
        }

    # Bearish FVG: third candle's high is below first candle's low
    if third["High"] < first["Low"]:
        return {
            "type": "BEARISH FVG",
            "lower": float(third["High"]),
            "upper": float(first["Low"])
        }

    return "NO FVG"


fvg_result = detect_fvg(frames["M15"])

print("GIVERS AI FVG ANALYSIS")
print("=====================")
print(fvg_result)
```

```python

# GIVERS AI - ENTRY ZONE FINDER

def find_entry_zone(df):

    candles = df.iloc[:-1].copy()

    if len(candles) < 105:
        return "INSUFFICIENT DATA"

    latest_gap = None

    # Search recent candles for FVGs
    start = len(candles) - 100

    for i in range(start, len(candles)):

        first = candles.iloc[i - 2]
        third = candles.iloc[i]

        # Bullish FVG
        if third["Low"] > first["High"]:

            lower = float(first["High"])
            upper = float(third["Low"])

            # Check whether the gap was fully filled later
            later = candles.iloc[i + 1:]

            if later.empty or later["Low"].min() > lower:
                latest_gap = {
                    "type": "BULLISH FVG",
                    "lower": lower,
                    "upper": upper
                }

        # Bearish FVG
        elif third["High"] < first["Low"]:

            lower = float(third["High"])
            upper = float(first["Low"])

            later = candles.iloc[i + 1:]

            if later.empty or later["High"].max() < upper:
                latest_gap = {
                    "type": "BEARISH FVG",
                    "lower": lower,
                    "upper": upper
                }

    if latest_gap is None:
        return "NO ACTIVE FVG ENTRY ZONE"

    return latest_gap


entry_zone = find_entry_zone(frames["M15"])

print("GIVERS AI ENTRY ZONE")
print("====================")
print(entry_zone)
```

```python

# GIVERS AI - RISK CALCULATOR

def calculate_risk(zone, df, risk_reward=2):

    if not isinstance(zone, dict):
        return "NO VALID ENTRY ZONE"

    candles = df.iloc[:-1].copy()
    last = candles.iloc[-1]

    entry = (zone["lower"] + zone["upper"]) / 2

    if zone["type"] == "BULLISH FVG":
        stop_loss = min(
            float(candles["Low"].iloc[-10:].min()),
            zone["lower"]
        )
        risk = entry - stop_loss
        take_profit = entry + (risk * risk_reward)

    else:
        stop_loss = max(
            float(candles["High"].iloc[-10:].max()),
            zone["upper"]
        )
        risk = stop_loss - entry
        take_profit = entry - (risk * risk_reward)

    if risk <= 0:
        return "INVALID RISK CALCULATION"

    return {
        "entry": round(entry, 6),
        "stop_loss": round(stop_loss, 6),
        "take_profit": round(take_profit, 6),
        "risk_reward": risk_reward
    }


risk_result = calculate_risk(entry_zone, frames["M15"])

print("GIVERS AI RISK ANALYSIS")
print("======================")
print(risk_result)
```
